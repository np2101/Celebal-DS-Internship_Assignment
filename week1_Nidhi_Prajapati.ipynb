### 4.3 Eigenvalues & Eigenvectors

For the matrix:
```
A = [[4, 1],
     [2, 3]]
```

1. Compute eigenvalues and eigenvectors using `np.linalg.eig`.
2. **Verify** the eigenvector equation: `A @ v = λ * v` for each eigenpair.
3. Plot both eigenvectors as arrows, scaled by their eigenvalues, on a 2D plot.
4. **Explain** in a markdown cell (below): what does it geometrically mean for a matrix to stretch a vector?

````python
import numpy as np
import matplotlib.pyplot as plt
A = np.array([[4, 1], [2, 3]], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors (columns):\n{eigenvectors}")

# Verify Av = λv for each eigenpair
for i in range(len(eigenvalues)):
    lam = eigenvalues[i]
    v   = eigenvectors[:, i]
    Av  = A @ v
    lv  = lam * v
    match = np.allclose(Av, lv)
    print(f"\nEigenpair {i+1}: λ={lam:.2f}, v={np.round(v,4)}")
    print(f"  A @ v = {np.round(Av, 4)}")
    print(f"  λ * v = {np.round(lv, 4)}  →  match: {match}")
    assert match, f"Eigenpair {i+1} failed verification!"

# Plot
fig, ax = plt.subplots()
colors = ['steelblue', 'tomato']
labels = [f'v₁ (λ={eigenvalues[0]:.1f})', f'v₂ (λ={eigenvalues[1]:.1f})']

for i in range(2):
    v     = eigenvectors[:, i]
    scale = eigenvalues[i]           # scale arrow by eigenvalue
    ax.quiver(0, 0, v[0]*scale, v[1]*scale,
              angles='xy', scale_units='xy', scale=1,
              color=colors[i], width=0.025, label=labels[i])
    ax.quiver(0, 0, v[0], v[1],      # original unit eigenvector (dashed)
              angles='xy', scale_units='xy', scale=1,
              color=colors[i], width=0.01, alpha=0.3)

ax.set_xlim(-4, 4); ax.set_ylim(-4, 4)
ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.set_title('Eigenvectors scaled by Eigenvalues')
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()

print("4.3 passed ✓")
````

**Your Geometric Explanation (4.3):**

>
When a matrix transforms a vector, it generally rotates and stretches it the output points in a completely different direction. But eigenvectors are special directions that the matrix only stretches or squishes, never rotates.

The eigenvector defines the direction that stays invariant under the transformation
The eigenvalue λ is the stretch factor along that direction:

λ > 1 → vector gets longer (stretched)

0 < λ < 1 → vector gets shorter (squished)

λ < 0 → vector flips direction and scales

λ = 1 → vector is completely unchanged



For matrix A above, λ=5 means the first eigenvector gets stretched to 5× its length, and λ=2 means the second gets stretched to 2× its length both staying exactly on their original lines.

### 4.4 SVD & Dimensionality Reduction

1. Create a random 4×3 matrix `X` (use `np.random.seed(42)`).
2. Compute its **SVD**: `U, S, Vt = np.linalg.svd(X, full_matrices=False)`.
3. **Reconstruct** `X` from the SVD and verify it matches the original.
4. Perform a **rank-1 approximation** using only the largest singular value/vector — call it `X_approx`.
5. Explain the relationship between SVD components and PCA in the markdown cell below.

````python
np.random.seed(42)
X = np.random.randn(4, 3)

U, S, Vt = np.linalg.svd(X, full_matrices=False)

X_reconstructed = U @ np.diag(S) @ Vt
X_approx        = S[0] * np.outer(U[:,0], Vt[0])

print(f"U shape: {U.shape}, S shape: {S.shape}, Vt shape: {Vt.shape}")
print(f"Singular values: {S}")
print(f"Reconstruction error: {np.linalg.norm(X - X_reconstructed):.2e}")
print(f"Rank-1 approximation:\n{np.round(X_approx, 4)}")

assert np.allclose(X, X_reconstructed, atol=1e-10), "Reconstruction failed!"
print("4.4 passed ✓")
````

**SVD → PCA Connection (4.4):**

> SVD decomposes any matrix X = U · Σ · Vᵀ into three parts:

| Component | Shape | Meaning |
|-----------|-------|---------|
| **U**     | (4×3) | Left singular vectors — how samples relate to components |
| **S (Σ)** | (3,)  | Singular values — magnitude of variance in each direction |
| **Vt**    | (3×3) | Right singular vectors — the principal component directions |

The relationship to PCA:

* Vt rows are the principal components — the directions of maximum
variance in feature space. This is exactly what PCA's eigenvectors give you
* Singular values S relate to eigenvalues by λᵢ = Sᵢ² / (n-1) — larger S means more variance explained
* U @ diag(S) gives the projected scores (coordinates of data in PCA space)

The rank-1 approximation keeps only the first singular value/vectors — the single direction that captures the most variance — giving the best possible 1D summary of the data. Each additional rank adds the next most important direction, which is exactly how PCA chooses components in order of decreasing variance.

### 5.4 Distribution Testing & Stationarity

1. Generate two samples: `s1` ~ Normal(0, 1), `s2` ~ Exponential(1), both n=200 (seed=0).
2. Run the **Kolmogorov-Smirnov test** (`stats.kstest`) comparing each against a standard normal CDF.
3. Generate a non-stationary time series (linear trend + noise). Run the **Augmented Dickey-Fuller test** (`adfuller` from `statsmodels`) and interpret the result.
4. Difference the series once and re-run ADF to confirm stationarity.

````python
from statsmodels.tsa.stattools import adfuller

np.random.seed(0)
s1 = np.random.normal(0, 1, 200)
s2 = np.random.exponential(1, 200)

# KS test — compare each against N(0,1)
ks_s1 = stats.kstest(s1, 'norm')
ks_s2 = stats.kstest(s2, 'norm')
print(f"KS test s1 (normal):      stat={ks_s1.statistic:.4f}, p={ks_s1.pvalue:.4f}")
print(f"KS test s2 (exponential): stat={ks_s2.statistic:.4f}, p={ks_s2.pvalue:.4f}")

# ADF test on non-stationary series
t  = np.arange(200)
ts = 0.05 * t + np.random.normal(0, 1, 200)

adf_result = adfuller(ts)
print(f"\nADF statistic: {adf_result[0]:.4f}")
print(f"ADF p-value:   {adf_result[1]:.4f}")
print(f"Stationary?    {adf_result[1] < 0.05}")

# Difference once and re-run
ts_diff  = np.diff(ts)
adf_diff = adfuller(ts_diff)
print(f"\nAfter differencing — p-value: {adf_diff[1]:.4f}, Stationary? {adf_diff[1] < 0.05}")
````

````python
def compute_psi(expected, actual, bins=10):
    """
    PSI = sum((actual_pct - expected_pct) * ln(actual_pct / expected_pct))
    Add epsilon=1e-10 to avoid log(0).
    """
    eps = 1e-10
    # Build bin edges from expected distribution
    breakpoints = np.linspace(np.min(expected), np.max(expected), bins + 1)

    # Count observations in each bin
    expected_counts = np.histogram(expected, bins=breakpoints)[0]
    actual_counts   = np.histogram(actual,   bins=breakpoints)[0]

    # Convert to percentages
    expected_pct = expected_counts / len(expected) + eps
    actual_pct   = actual_counts   / len(actual)   + eps

    # PSI formula
    psi = np.sum((actual_pct - expected_pct) * np.log(actual_pct / expected_pct))
    return psi

np.random.seed(1)
train_dist = np.random.normal(50, 10, 1000)   # training distribution
drift_dist = np.random.normal(65, 12, 1000)   # shifted production distribution

psi_value = compute_psi(train_dist, drift_dist)
print(f"PSI: {psi_value:.4f}")
print(f"Shift severity: {'Major' if psi_value > 0.2 else 'Minor' if psi_value > 0.1 else 'Stable'}")

# Plot both distributions
fig, ax = plt.subplots()
ax.hist(train_dist, bins=30, alpha=0.6, color='steelblue',
        density=True, label='Training (μ=50, σ=10)')
ax.hist(drift_dist, bins=30, alpha=0.6, color='tomato',
        density=True, label='Production (μ=65, σ=12)')

# KDE overlays
for dist, color in [(train_dist, 'steelblue'), (drift_dist, 'tomato')]:
    kde = stats.gaussian_kde(dist)
    x   = np.linspace(dist.min(), dist.max(), 300)
    ax.plot(x, kde(x), color=color, lw=2)

ax.legend(); ax.set_title(f'Distribution Shift Visualization  (PSI={psi_value:.3f})')
ax.set_xlabel('Value')
plt.tight_layout(); plt.show()
````

**Concept Drift vs Covariate Drift & PSI Thresholds (5.5):**

- **Concept drift:** The relationship between input features and the target variable changes over time — e.g. a fraud detection model fails because fraudsters adopt new behaviour patterns not seen during training.
- **Covariate drift:** The distribution of input features P(X) shifts between training and production, even if P(Y|X) stays the same — e.g. a model trained on young users starts receiving data mostly from older users.
- **PSI < 0.1:** Stable — no significant shift detected; model can continue in production without changes.
- **PSI 0.1–0.2:** Minor shift — monitor the distribution closely and schedule a review; retraining may soon be needed.
- **PSI > 0.2:** Major shift — distribution has changed significantly; investigate immediately and retrain the model on recent data.
- **Retraining trigger example:** When PSI exceeds 0.2 on any key input feature for two consecutive monitoring cycles, automatically trigger the retraining pipeline using the last 90 days of production data.

### 6.1 Core Concepts

A bag contains: 4 red, 3 blue, 3 green marbles.

1. Define the sample space and compute P(red), P(blue), P(green).
2. Two marbles drawn **without replacement**. Compute:
   - P(first=red, second=blue) — **joint probability**
   - P(second=blue | first=red) — **conditional probability**
3. Are the two draws **independent**? Show mathematically.

````python
total = 10
red, blue, green = 4, 3, 3

p_red   = red   / total        # 4/10
p_blue  = blue  / total        # 3/10
p_green = green / total        # 3/10

# Joint: P(red first AND blue second) = P(red) * P(blue|red)
p_red_then_blue  = (red / total) * (blue / (total - 1))   # 4/10 * 3/9

# Conditional: P(blue second | red first) = 3 blue remaining / 9 total remaining
p_blue_given_red = blue / (total - 1)                     # 3/9

# Independence check: draws are independent only if P(blue|red) == P(blue)
independent = np.isclose(p_blue_given_red, p_blue)

print(f"P(red)={p_red:.3f}, P(blue)={p_blue:.3f}, P(green)={p_green:.3f}")
print(f"P(red,blue): {p_red_then_blue:.4f}")
print(f"P(blue|red): {p_blue_given_red:.4f}")
print(f"P(blue):     {p_blue:.4f}")
print(f"Independent? {independent}")
````

### 6.3 Bayes' Theorem

**Scenario — Spam filter:**
- P(Spam) = 0.30 (prior)
- P("free" | Spam) = 0.80 (likelihood)
- P("free" | Not Spam) = 0.05

1. Compute P(Spam | "free") using Bayes' theorem.
2. Implement a simple `naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham)` that returns the posterior P(Spam | word).
3. In the markdown cell, map each term — **prior, likelihood, evidence, posterior** — to the spam example.

````python
p_spam            = 0.30
p_free_given_spam = 0.80
p_free_given_ham  = 0.05
p_ham             = 1 - p_spam

# P(free) — law of total probability
p_free = (p_free_given_spam * p_spam) + (p_free_given_ham * p_ham)

# Bayes
p_spam_given_free = (p_free_given_spam * p_spam) / p_free

print(f"P('free'): {p_free:.4f}")
print(f"P(Spam | 'free'): {p_spam_given_free:.4f}")

def naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham):
    prior_ham = 1 - prior_spam
    evidence = (p_word_given_spam * prior_spam) + (p_word_given_ham * prior_ham)
    posterior = (p_word_given_spam * prior_spam) / evidence
    return posterior

pred = naive_bayes_predict(0.30, 0.80, 0.05)
assert abs(pred - p_spam_given_free) < 1e-9
print(f"naive_bayes_predict: {pred:.4f}")
print("6.3 passed ✓")
````

### 6.4 Central Limit Theorem

1. Define a population following an **exponential distribution** (λ=1, size=100,000).
2. Repeatedly draw samples of size `n=30` and record their means — do this 5,000 times.
3. Plot the histogram of sample means.
4. Overlay the **theoretical normal distribution** predicted by the CLT: μ = population mean, σ = population_std / sqrt(n).
5. Run a KS test to confirm the distribution of sample means is approximately normal.

````python
np.random.seed(7)
population = np.random.exponential(scale=1.0, size=100_000)

n_samples   = 5000
sample_size = 30
sample_means = np.array([
    np.random.choice(population, size=sample_size, replace=False).mean()
    for _ in range(n_samples)
])

pop_mean = population.mean()
pop_std  = population.std()
clt_std  = pop_std / np.sqrt(sample_size)   # σ / √n

print(f"Population mean: {pop_mean:.4f}, Population std: {pop_std:.4f}")
print(f"Sample means mean: {sample_means.mean():.4f}, std: {sample_means.std():.4f}")
print(f"CLT predicted std: {clt_std:.4f}")

# Plot
fig, ax = plt.subplots(figsize=(8, 5))

# Histogram of sample means (density=True so it aligns with the PDF overlay)
ax.hist(sample_means, bins=50, density=True, alpha=0.6,
        color='steelblue', edgecolor='white', label='Sample means (n=30)')

# Theoretical normal curve predicted by CLT
x = np.linspace(sample_means.min(), sample_means.max(), 300)
ax.plot(x, stats.norm.pdf(x, loc=pop_mean, scale=clt_std),
        color='crimson', linewidth=2.5,
        label=f'CLT normal  μ={pop_mean:.3f}, σ={clt_std:.3f}')

ax.set_xlabel('Sample Mean')
ax.set_ylabel('Density')
ax.set_title('CLT: Distribution of Sample Means (Exponential Population)')
ax.legend()
plt.tight_layout()
plt.show()

# KS test
ks_result = stats.kstest(sample_means, 'norm', args=(pop_mean, clt_std))
print(f"KS test p-value: {ks_result.pvalue:.4f} → Approximately normal? {ks_result.pvalue > 0.05}")
````

**CLT Reflection (6.4):**

> The CLT matters because it guarantees that sample means become normally distributed regardless of the population's shape, making normal-based inference valid even for skewed distributions. This justifies the normality assumption in classical tests like z-tests, t-tests, and confidence intervals they require the estimator to be normal, not the raw data. In ML, this explains why batch loss averages and gradient updates behave predictably at sufficient sample sizes, enabling stable optimization over non-Gaussian data.
